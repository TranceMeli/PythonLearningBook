## 21. How to parse a string into a datetime object?
**Stack Overflow:** [View Post](https://stackoverflow.com/questions/466345/how-to-parse-a-string-into-a-datetime-object)

**Problem:** The user wants to convert a formatted date string into a Python `datetime` object.

**Answer:** Use `datetime.strptime()` specifying the appropriate format directives.

```python
from datetime import datetime
date_str = "2026-10-13 14:30:00"
dt_obj = datetime.strptime(date_str, "%Y-%m-%d %H:%M:%S")
print(dt_obj, type(dt_obj))
```

## 25. How do I sort a list of dictionaries in Python?
**Stack Overflow:** [View Post](https://stackoverflow.com/questions/72899/how-do-i-sort-a-list-of-dictionaries-in-python)

**Problem:** The user wants to sort a list containing dictionary objects based on a specific key value.

**Answer:** Use `sorted()` combined with a lambda function.

```python
users = [{"name": "Alice", "age": 30}, {"name": "Bob", "age": 25}]
sorted_users = sorted(users, key=lambda x: x["age"])
print(sorted_users)
```

## 26. How to take the union of multiple lists (unique values)?
**Stack Overflow:** [View Post](https://stackoverflow.com/questions/1230104/how-to-take-union-of-multiple-lists-in-python)

**Problem:** The user wants to combine multiple lists while removing duplicate values.

**Answer:** Convert the lists to `set`s and use the union operator `|`.

```python
list1 = [1, 2, 3]
list2 = [3, 4, 5]
unique_union = list(set(list1) | set(list2))
print(unique_union)
```

## 27. How to reverse a list in Python?

```python
items = [1, 2, 3, 4]
reversed_items = items[::-1]
print(reversed_items)
```

## 29. How do I check if all items in a list match a condition?
**Stack Overflow:** [View Post](https://stackoverflow.com/questions/1524350/how-do-i-check-if-all-items-in-a-list-match-a-condition)

**Problem:** The user wants to test if every element in an iterable satisfies a boolean condition.

**Answer:** Use the built-in `all()` function with a generator expression.

```python
numbers = [2, 4, 6, 8]
all_even = all(n % 2 == 0 for n in numbers)
print("All even:", all_even)
```

## 31. Random string generation with upper-case letters and digits
**Stack Overflow:** [View Post](https://stackoverflow.com/questions/2257441/random-string-generation-with-upper-case-letters-and-digits)

**Problem:** The user wants to generate random alphanumeric tokens or strings.

**Answer:** Combine `random.choices` with `string.ascii_letters` and `string.digits`.

```python
import random
import string
token = "".join(random.choices(string.ascii_letters + string.digits, k=10))
print(token)
```

## 32. How do I remove duplicates from a list while preserving order?
**Stack Overflow:** [View Post](https://stackoverflow.com/questions/480214/how-do-i-remove-duplicates-from-a-list-while-preserving-order)

**Problem:** The user wants to strip duplicate entries from a list without scrambling the original item order.

**Answer:** Use `dict.fromkeys()` which preserves insertion order.

```python
duplicates = [1, 2, 2, 3, 1, 4]
unique_list = list(dict.fromkeys(duplicates))
print(unique_list)
```
