#### Variables and Types<br>
Python is a **dynamically typed** language: you don't need to declare a variable's type before using it, and a variable can be reassigned to a different type later. Python figures out the type automatically at runtime.<br>
Separately, Python is also **object-oriented**: every value you create — even a simple number or string — is internally represented as an object. These are two different concepts, but they're often mentioned together because both mean you rarely have to think about low-level type declarations.

```python
x = "Python is awesome"
print(x)
```

#### Python Data Types

#### Data Types

| Category | Types |
|---|---|
| Text Type | `str` |
| Numeric Types | `int`, `float`, `complex` |
| Sequence Types | `list`, `tuple`, `range` |
| Mapping Type | `dict` |
| Set Types | `set`, `frozenset` |

```python
# Text Type
x = "Hello World"

# Numeric Types
x = 20          # int
x = 20.5        # float
x = 1j          # complex

# Sequence Types
x = ["apple", "banana", "cherry"]   # list
x = ("apple", "banana", "cherry")   # tuple
x = range(6)                        # range

# Mapping Type
x = {"name": "John", "age": 36}     # dict

# Set Types
x = {"apple", "banana", "cherry"}            # set
x = frozenset({"apple", "banana", "cherry"}) # frozenset

# Boolean Type
x = True

# Binary Types
x = b"Hello"                  # bytes
x = bytearray(5)              # bytearray
x = memoryview(bytes(5))      # memoryview

# None Type
x = None
```

#### Python Operators

#### Arithmetic Operators

| Operator | Name | Example |
|---|---|---|
| `+` | Addition | `x + y` |
| `-` | Subtraction | `x - y` |
| `*` | Multiplication | `x * y` |
| `/` | Division | `x / y` |
| `%` | Modulus | `x % y` |
| `**` | Exponentiation | `x ** y` |
| `//` | Floor division | `x // y` |

```python
x = 10
y = 3

print(x + y)   # Addition -> 13
print(x - y)   # Subtraction -> 7
print(x * y)   # Multiplication -> 30
print(x / y)   # Division -> 3.3333333333333335
print(x % y)   # Modulus -> 1
print(x ** y)  # Exponentiation -> 1000
print(x // y)  # Floor division -> 3
```

#### Assignment Operators

Assignment operators are used to assign values to variables.

| Operator | Example | Same As |
|---|---|---|
| `=` | `x = 5` | `x = 5` |
| `+=` | `x += 3` | `x = x + 3` |
| `-=` | `x -= 3` | `x = x - 3` |
| `*=` | `x *= 3` | `x = x * 3` |
| `/=` | `x /= 3` | `x = x / 3` |
| `%=` | `x %= 3` | `x = x % 3` |
| `//=` | `x //= 3` | `x = x // 3` |
| `**=` | `x **= 3` | `x = x ** 3` |
| `&=` | `x &= 3` | `x = x & 3` |
| `\|=` | `x \|= 3` | `x = x \| 3` |
| `^=` | `x ^= 3` | `x = x ^ 3` |
| `>>=` | `x >>= 3` | `x = x >> 3` |
| `<<=` | `x <<= 3` | `x = x << 3` |
| `:=` | `print(x := 3)` | `x = 3`<br>`print(x)` |

```python
# Basic Assignment
x = 5
print(x)        # 5

# Addition Assignment
x = 5
x += 3
print(x)        # 8

# Subtraction Assignment
x = 5
x -= 3
print(x)        # 2

# Multiplication Assignment
x = 5
x *= 3
print(x)        # 15

# Division Assignment
x = 5
x /= 3
print(x)        # 1.6666666666666667

# Modulus Assignment
x = 5
x %= 3
print(x)        # 2

# Floor Division Assignment
x = 5
x //= 3
print(x)        # 1

# Exponentiation Assignment
x = 5
x **= 3
print(x)        # 125

# Bitwise AND Assignment
x = 5
x &= 3
print(x)        # 1

# Bitwise OR Assignment
x = 5
x |= 3
print(x)        # 7

# Bitwise XOR Assignment
x = 5
x ^= 3
print(x)        # 6

# Bitwise Right Shift Assignment
x = 5
x >>= 3
print(x)        # 0

# Bitwise Left Shift Assignment
x = 5
x <<= 3
print(x)        # 40

# Walrus Operator (assign and return in one step)
print(x := 3)   # 3
print(x)        # 3
```

#### Ternary Operator

A one-line shorthand for an `if-else` statement.

| Syntax | Example |
|---|---|
| `x if condition else y` | `"Yes" if a > b else "No"` |

```python
a = 330
b = 200
print("A") if a > b else print("B")   # A

# Common use: assign a value conditionally
result = "Even" if a % 2 == 0 else "Odd"
print(result)   # Even
```

#### Comparison Operators

Used to compare two values.

| Operator | Name | Example |
|---|---|---|
| `==` | Equal | `x == y` |
| `!=` | Not equal | `x != y` |
| `>` | Greater than | `x > y` |
| `<` | Less than | `x < y` |
| `>=` | Greater than or equal to | `x >= y` |
| `<=` | Less than or equal to | `x <= y` |

```python
x = 5
y = 3

print(x == y)   # False
print(x != y)   # True
print(x > y)    # True
print(x < y)    # False
print(x >= y)   # True
print(x <= y)   # False
```

#### Logical Operators

Used to combine conditional statements.

| Operator | Description | Example |
|---|---|---|
| `and` | True if both statements are true | `x < 5 and x < 10` |
| `or` | True if at least one statement is true | `x < 5 or x < 4` |
| `not` | Reverses the result | `not(x < 5 and x < 10)` |

```python
x = 5

print(x < 10 and x > 0)       # True
print(x < 4 or x > 3)         # True
print(not(x < 10 and x > 0))  # False
```

## Python Collections

<b>List</b> is a collection which is ordered and changeable. Allows duplicate members<br>
<b>Tuple</b> is a collection which is ordered and unchangeable. Allows duplicate members.<br>
<b>Set</b> is a collection which is unordered, unchangeable, and unindexed. No duplicate members<br>
<b>Dictionary</b> is a collection which is ordered andd changeable. No duplicate members.

#### Python Lists

```python
# list items are ordered, changeable and allow duplicated values
mylist = ["apple", "banana", "cherry"]
print(mylist)
```

#### Python Tuple

```python
# tuple
# are used to store multiple items in a single variable
# ordered and unchangeable
mytuple = ("apple", "banana", "cherry", "banana")
print(mytuple)
```

ordered, unchangeable and allow duplicate<br>
ordered, items have defined order and the order will not change<br>
unchangeable, cannot change, add or remove items after the tuple has been created

#### Python Set

```python
# sets
# a collection which is unordered, unchangeable, and unindexed
myset = {"apple", "banana", "cherry"}
print(myset)
```

unordered<br>
set items can appear in a different order every time you use them, and cannot be referred to by index or key<br>
unchangeable<br>

#### Python Conditions and If statements

```python
a = 33
b = 200
if b > a:
    print("b is greater than a")
```

If the condition is true, the code block inside the if statement is executed. If the condition is false, the code block is skipped.

```python
# elif
# if the previous conditions were not true, then try this condition
a = 33
b = 33
if b > a:
    print("b is greater than a")
elif a == b:
    print("a and b are equal")
```

```python
a = 200
b = 33
if b > a:
    print("b is greater than a")
elif a == b:
    print("a and b are equal")
else:
    print("a is greater than b")
```

# Python loops

python has two loops, `while` and `for` loops

```python
# print i as long as i is less than 6
i = 1
while i < 6:
    print(i)
    i += 1
```

with the `break` statement we can stop the loop

```python
i = 1
while i < 6:
  print(i)
  if (i == 3):
    break
  i += 1
```

with `continue` we can stop the curret iteration

```python
i = 0
while i < 6:
  i += 1
  if i == 3:
    continue
  print(i)
```

with else we can run a block of code once when the condition is no longer true

```python
i = 1
while i < 6:
  print(i)
  i += 1
else:
  print("i is no longer less than 6")
```
