## 41. What is the difference between `copy.copy()` and `copy.deepcopy()`?
**Stack Overflow:** [View Post](https://stackoverflow.com/questions/17240/what-is-the-difference-between-shallow-and-deep-copy)

**Problem:** The user wants to understand how shallow and deep copying differ when duplicating complex objects.

**Answer:** A shallow copy constructs a new collection object and populates it with references to the child objects, whereas a deep copy recursively duplicates all nested objects.

```python
import copy
original = [[1, 2], [3, 4]]
shallow = copy.copy(original)
deep = copy.deepcopy(original)
original[0][0] = 99
print("Shallow reflects change:", shallow[0][0])
print("Deep preserves original:", deep[0][0])
```

## 42. How to flatten a list of lists using itertools?
**Stack Overflow:** [View Post](https://stackoverflow.com/questions/952914/how-to-make-a-flat-list-out-of-list-of-lists)

**Problem:** The user wants an efficient way to flatten nested iterables using the standard library.

## 43. How to split a list into evenly sized chunks?
**Stack Overflow:** [View Post](https://stackoverflow.com/questions/312443/how-do-you-split-a-list-into-evenly-sized-chunks)

**Problem:** The user wants to partition a list into smaller chunks of a specified size.

**Answer:** Use list comprehension with slicing or generator expressions.

```python
def chunk_list(lst, chunk_size):
    return [lst[i:i + chunk_size] for i in range(0, len(lst), chunk_size)]

data = [1, 2, 3, 4, 5, 6, 7]
print(chunk_list(data, 3))
```

## 45. What is the difference between `list.sort()` and `sorted()`?
**Stack Overflow:** [View Post](https://stackoverflow.com/questions/2244033/whats-the-difference-between-sort-and-sorted)

**Problem:** The user wants to know when to use the list `.sort()` method versus the built-in `sorted()` function.

**Answer:** `list.sort()` sorts a list *in-place* and returns `None`, whereas `sorted()` accepts any iterable and returns a *new* sorted list.

```python
numbers = [3, 1, 4, 1, 5]
new_sorted = sorted(numbers)  # returns new list
numbers.sort()                # sorts in-place
print("Sorted copy:", new_sorted)
print("Original sorted in-place:", numbers)
```

## 47. Does Python have a switch / case statement?
**Stack Overflow:** [View Post](https://stackoverflow.com/questions/60208/replacements-for-switch-case-statement-in-python)
**Problem:** The user is looking for a clean conditional multi-branch selection structure.

**Answer:** Since Python 3.10, structural pattern matching (`match` / `case`) is natively available.

```python
def http_status(status):
    match status:
        case 200:
            return "OK"
        case 404:
            return "Not Found"
        case _:
            return "Unknown"

print(http_status(200))
```

## 48. How to iterate over two or more lists simultaneously (`zip`)?

## 52. Difference between `del`, `remove()`, and `pop()` on lists
**Stack Overflow:** [View Post](https://stackoverflow.com/questions/11520492/difference-between-del-remove-and-pop-on-lists)
**Problem:** The user wants to understand the different ways to delete elements from lists.

**Answer:** `remove()` deletes by value, `pop()` removes by index and returns the item, and `del` deletes by index or slice without returning.

```python
lst = [10, 20, 30, 40]
lst.remove(20)  # removes value 20
val = lst.pop(0)  # removes and returns index 0
del lst[0]      # deletes index 0
print(lst)
```

## 53. How to check if a string represents a number?
**Stack Overflow:** [View Post](https://stackoverflow.com/questions/354038/how-do-i-check-if-a-string-is-a-number-float)
**Problem:** The user wants to validate whether string input can be parsed as a numeric value.

**Answer:** Use string methods like `.isdigit()` for integers, or try parsing with `float()` inside a helper function.

```python
def is_number(s):
    try:
        float(s)
        return True
    except ValueError:
        return False

print(is_number("123.45"))
print(is_number("abc"))
```

## 57. How to measure elapsed execution time in Python?
**Stack Overflow:** [View Post](https://stackoverflow.com/questions/739684/how-to-report-execution-time-in-python)
**Problem:** The user wants to benchmark code runtime using the standard library.
**Answer:** Use `time.perf_counter()` for high-resolution timing measurements.

```python
import time
start_time = time.perf_counter()
# Do some work
sum(range(100000))
end_time = time.perf_counter()
print(f"Elapsed time: {end_time - start_time:.6f} seconds")
```

## 60. How to create custom exceptions in Python?
**Stack Overflow:** [View Post](https://stackoverflow.com/questions/1319615/proper-way-to-declare-custom-exceptions-in-python)
**Problem:** The user wants to define domain-specific error types for exception handling.
**Answer:** Inherit a new class from the built-in `Exception` base class.

```python
class ValidationError(Exception):
    pass

try:
    raise ValidationError("Invalid configuration data!")
except ValidationError as e:
    print("Caught custom exception:", e)
```
