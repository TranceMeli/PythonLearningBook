# Classes and Objects

A class defines what an object should look like. You can call it a blueprint. An object is created based on that class.<br><br>

Class   Objects<br>
Fruit   Apple, Banana, Mango

```python
# define a class
class Dog:
    sound = "bark"

# creating a object from class
dog1 = Dog()
print(dog1.sound)
```

__init__() method acts as a constructor

```python
class Dog:
    species = "Canine" # class attribute

    def __init__(self,name, age):
        self.name = name # instance attribute
        self.age = age # instance attribute

# creating an object of the dog class
dog1 = Dog("Buddy", 3)

print(dog1.name)
print(dog1.species)
```

```python
class Dog:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __str__(self):
        return f"{self.name} is {self.age} years old."

dog1 = Dog("Buddy", 3)
dog2 = Dog("Charlie", 5)

print(dog1)
print(dog2)

    
```
