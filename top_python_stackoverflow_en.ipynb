## 1. What does the "yield" keyword do?
**Stack Overflow:** [View Post](https://stackoverflow.com/questions/231767/what-does-the-yield-keyword-do)

**Problem:** The user wants to understand how generator functions work and what the purpose of the `yield` keyword is compared to a standard `return` statement.

**Answer:** `yield` pauses the function's execution, saves its state, and returns a value to the caller. When resumed, execution continues right where it left off.

```python
def simple_generator():
    yield 1
    yield 2
    yield 3

gen = simple_generator()
print(next(gen))
print(next(gen))
```

## 10. What is the difference between `@staticmethod` and `@classmethod`?
**Stack Overflow:** [View Post](https://stackoverflow.com/questions/1217927/what-is-the-difference-between-staticmethod-and-classmethod)

**Problem:** The user wants to understand when to use `@classmethod` versus `@staticmethod` inside classes.

**Answer:** `@classmethod` takes the class (`cls`) as its first implicit argument; `@staticmethod` takes neither instance (`self`) nor class (`cls`), acting like a regular function scoped inside the class namespace.

```python
class MyClass:
    @classmethod
    def class_method(cls):
        return f"Class: {cls.__name__}"
    
    @staticmethod
    def static_method():
        return "No class or instance data needed"

print(MyClass.class_method())
print(MyClass.static_method())
```

## 13. How do I sort a dictionary by value?
**Stack Overflow:** [View Post](https://stackoverflow.com/questions/613183/how-do-i-sort-a-dictionary-by-value)

**Problem:** The user wants to sort dictionary items based on their values rather than keys.

**Answer:** Use `sorted()` with a lambda key targeting values, then reconstruct the dictionary.

```python
scores = {"Alice": 85, "Bob": 92, "Charlie": 78}
sorted_scores = dict(sorted(scores.items(), key=lambda item: item[1], reverse=True))
print(sorted_scores)
```

## 15. What is the difference between append() and extend()?
**Stack Overflow:** [View Post](https://stackoverflow.com/questions/252703/what-is-the-difference-between-append-and-extend)

**Problem:** The user wants to understand how `append()` and `extend()` differ when adding elements to lists.

**Answer:** `append()` adds its argument as a single element; `extend()` iterates over its argument and adds each element individually.

```python
a = [1, 2]
a.append([3, 4])
print("append:", a)

b = [1, 2]
b.extend([3, 4])
print("extend:", b)
```

## 16. Difference between __str__ and __repr__
**Stack Overflow:** [View Post](https://stackoverflow.com/questions/1436703/difference-between-str-and-repr)

**Problem:** The user wants to know the distinct purposes of `__str__` and `__repr__` special methods.

**Answer:** `__str__` provides an informal, user-friendly string representation, while `__repr__` provides an unambiguous representation intended for developers and debugging.

```python
class Car:
    def __init__(self, brand):
        self.brand = brand
    def __str__(self):
        return f"Car of brand {self.brand}"
    def __repr__(self):
        return f"Car(brand='{self.brand}')"

c = Car("BMW")
print(str(c))
print(repr(c))
```

## 18. What do *args and **kwargs mean?
**Stack Overflow:** [View Post](https://stackoverflow.com/questions/36901/what-does-args-and-kwargs-mean)

**Problem:** The user wants to understand how to handle a variable number of positional and keyword arguments in functions.

**Answer:** `*args` collects extra positional arguments into a tuple, and `**kwargs` collects extra keyword arguments into a dictionary.

```python
def flex_func(*args, **kwargs):
    print("Args:", args)
    print("Kwargs:", kwargs)

flex_func(1, 2, 3, name="Python", type="Language")
```
